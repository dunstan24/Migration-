```python
import pandas as pd

# Tambahkan kolom penanda asal file
df_occ["Source"] = "File A"
df_occ2["Source"] = "File B"

# Gabungkan
combined = pd.concat([df_occ, df_occ2])

# Hapus duplikat berdasarkan semua kolom KECUALI kolom 'Source'
# Ini akan menyisakan baris yang isinya berbeda
columns_to_check = [col for col in combined.columns if col != "Source"]
diff = combined.drop_duplicates(subset=columns_to_check, keep=False)

if diff.empty:
    print("Kedua file memiliki data yang identik!")
else:
    print(f"Ditemukan {len(diff)} baris yang berbeda:")
    # Urutkan agar perbedaan yang mirip terlihat berdekatan
    print(diff.sort_values(by=columns_to_check))
```
